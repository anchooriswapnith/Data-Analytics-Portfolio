```python
df['Review Rating'] = df.groupby('Category')['Review Rating'].transform(lambda x: x.fillna(x.mean()))
```

```python
df.isnull().sum()
```

```python
df.columns=df.columns.str.lower()
df.columns = df.columns.str.replace(" ","_")
df.columns
```

```python
labels = ["Young Adult","Adult","Middel-aged","Senior"]
df['age groups'] = pd.qcut(df['age'],q=4,labels=labels)
```

```python
df.loc[df['age'] ==55,['age','age groups']]
```

```python
frequency_mapping = {"Fortnightly": 14,
                     "Weekly":7,
                     "Monthly":30,
                     "Quarterly":90,
                     "Bi-Weekly":14,
                     "Annually":365,
                     "Every 3 Months":90}
df['purchase_frequency_days']=df['frequency_of_purchases'].map(frequency_mapping)
```

```python
df["purchase_frequency_days"].isnull().sum()
```
